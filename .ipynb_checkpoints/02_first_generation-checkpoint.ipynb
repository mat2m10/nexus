# Import libraries

```python
import json
import numpy as np
import pandas as pd
from helpers.phenotype import attach_populations
```

# Load params

```python
with open("params.json") as f:
    params = json.load(f)
k           = params["k"]
Ne          = params["Ne"]
M           = params["M"]
n_per_deme  = params["n_per_deme"]
n_loci      = params["n_loci"]
n_mutations = params["n_mutations"]
random_seed = params["random_seed"]
```

# Load data

```python
fname = f"data/G_k{k}_Ne{Ne}_M{M}_npd{n_per_deme}_nloc{n_loci}_seed{random_seed}.npz"
data  = np.load(fname)
G     = data["G"]
```

```python
G
```

```python
humans = attach_populations(G.shape[1], k, n_per_deme)
```

```python
G_T = G.T

results = {}
pops = humans['populations'].values  # length 2000

for pop in np.unique(pops):
    mask = pops == pop
    G_pop = G_T[mask]           # subset individuals for this population
    freq = G_pop.mean(axis=0)   # frequency of 1s per SNP
    results[pop] = freq

freq_df = pd.DataFrame(results).T
# Shape: (n_populations x 100 SNPs)
```

```python
n_pops, n_snps = freq_df.shape

simulated_list = []

for pop in freq_df.index:
    freqs = freq_df.loc[pop].values          # (100,) frequencies for this pop
    # Draw n_individuals genotypes, each SNP independently
    pop_genotypes = np.random.binomial(n=1, p=freqs, size=(n_per_deme, n_snps))
    pop_df = pd.DataFrame(pop_genotypes, columns=freq_df.columns)
    for col in pop_df.columns:
        idx = np.random.randint(0, len(pop_df))
        pop_df.loc[idx, col] = 1 - pop_df.loc[idx, col]
    pop_df['population'] = pop
    simulated_list.append(pop_df)

simulated_df = pd.concat(simulated_list, ignore_index=True)
```

```python
simulated_df
```

```python
cols = simulated_df.columns.difference(['population'])

freq_df = simulated_df.groupby('population')[cols].mean()
```

```python
def generate_genotypes(freq_df, F, n_individuals):
    dfs = []
    
    for pop, row in freq_df.iterrows():
        q = row.values                        # minor allele freq, shape (n_snps,)
        p = 1 - q

        prob_major = p**2 + F * p * q
        prob_het   = 2 * p * q * (1 - F)

        # draw n_individuals x n_snps random numbers
        r = np.random.uniform(size=(n_individuals, len(q)))

        genotypes = np.where(
            r < prob_major, 1,
            np.where(r < prob_major + prob_het, 0, -1)
        )

        pop_df = pd.DataFrame(genotypes, columns=freq_df.columns)
        pop_df['population'] = pop
        dfs.append(pop_df)
    
    genotype_df = pd.concat(dfs, ignore_index=True)
    return genotype_df

genotype_df = generate_genotypes(freq_df, 0.1, n_per_deme)
```

```python
genotype_df
```

```python
snp_cols = [col for col in genotype_df.columns if col != 'population']

allele_counts = (genotype_df[snp_cols] == -1) * 2 + (genotype_df[snp_cols] == 0) * 1
maf = allele_counts.mean(axis=0) / 2

# Flip columns where MAF > 0.5
cols_to_flip = maf[maf > 0.5].index
genotype_df[cols_to_flip] = genotype_df[cols_to_flip] * -1

# Recompute MAF after flipping
allele_counts = (genotype_df[snp_cols] == -1) * 2 + (genotype_df[snp_cols] == 0) * 1
maf = allele_counts.mean(axis=0) / 2

def maf_prefix(m):
    if m < 0.01:
        return "VR"
    elif m < 0.05:
        return "R"
    else:
        return "C"

sorted_cols = maf.sort_values().index
new_columns = {
    col: f"{maf_prefix(maf[col])}_{maf[col]:.2f}_snp_{i+1}"
    for i, col in enumerate(sorted_cols)
}

genotype_df = genotype_df[list(sorted_cols) + ['population']].rename(columns=new_columns)
```
